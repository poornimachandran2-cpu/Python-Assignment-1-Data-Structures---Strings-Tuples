# ***Strings (Concatenation, Slicing and Other methods) :***


**1. String Concatenation:**

String concatenation is the process of joining two or more strings using the + operator. In this program, the user's name is combined with the greeting string "Hello " and later another message is appended to create a complete sentence.

```python
string1 = "Hello "
name = input("enter your name : ")
result1 = string1 + name
print(result1)
```

```python
string2 =  " , Welcome to Python programming"
result2 = result1 + string2
print(result2)
```

**2. String Slicing and Indexing:**

String indexing is used to access individual characters in a string, while slicing is used to extract a portion of the string. Python uses zero-based indexing, where the first character is at position 0.

```python
# a. Print the first character of the string
print(result2[0])
```

```python
# b. Print the last character of the string
print(result2[-1])
```

```python
# c. Print the first 5 characters of the string
print(result2[0:5])
```

```python
# d. Print the last 11 characters of the string
print(result2[-11:])
```

```python
# e. Print the string in reverse
print(result2[::-1])
```

```python
# f. Use slicing and print the word “Python” from the existing string
print(result2[25:32])
```

**3. String Methods:**

This is built in string method in Python like converting text to uppercase, lowercase, counting characters, and replacing words.

```python
# a. Convert the sentence to uppercase.
strM = "Python beginner tutorial"
print(strM.upper())
```

```python
# b. Convert the sentence to lowercase
print(strM.lower())
```

```python
# c. Use Capitalize and return the sentence to the original input form.
print(strM.capitalize())
```

```python
# d. Count the total number of occurrences of character ‘t’ in the string
print("count of 't':", strM.count('t'))
```

```python
# e. Replace all occurrences of “Python” with “Machine Learning” in the input string
strM = "Python beginner tutorial"
newstrM =strM.replace("Python","Machine Learning")
print(newstrM)
```

# ***Tuples (Creation, Modification and Access) :***

Tuple is an ordered collection of elements. They are immutable.

```python
# Create 1st tuple with values -> (10, 20, 30)
tuple1 = (10,20,30)
```

```python
# 2nd tuple with values -> (40, 50, 60)
tuple2 = (40,50,60)
```

```python
# a. Concatenate the two tuples and store it in “t_combine”
t_combine = tuple1 + tuple2
print(t_combine)
```

```python
# b. Repeat the elements of “t_combine” 3 times
print(t_combine * 3)
```

```python
# c. Access the 3rd element from “t_combine”
print(t_combine[2])
```

```python
# d. Access the first three elements from “t_combine”
print(t_combine[:3])
```

```python
# e. Access the last three elements from “t_combine”
print(t_combine[-3:])
```
